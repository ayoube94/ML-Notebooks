#✨ Capa Convolucional y Redes Neuronales Convolucionales (CNNs)

Las Redes Neuronales Convolucionales (CNNs) son un tipo de red neuronal diseñada para procesar imágenes. Se componen principalmente de capas convolucionales, capas de agrupación (pooling layers) y capas densas.

##📃 Configuración del Entorno en Google Colab

Para ejecutar este código en Google Colab, primero asegurémonos de instalar TensorFlow y cargar las bibliotecas necesarias.

```python
# Verificar e instalar TensorFlow si es necesario
import tensorflow as tf
from tensorflow import keras

from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, AveragePooling2D
```

###🔍 1. Capa Convolucional

📊 Concepto

La capa convolucional extrae patrones de las imágenes usando filtros (kernels) que recorren la imagen y detectan características como bordes, texturas y formas.

🔢 Cómo funciona

Entrada: La imagen se representa como una matriz de píxeles (ejemplo: 28x28 para escala de grises o 32x32x3 para RGB).

Operación de convolución:

1. Filtro (Kernel): Una pequeña matriz (ej. 3x3 o 5x5) se desliza sobre la imagen.

2. Se multiplican los valores del filtro por los píxeles correspondientes.

3. Se suman los resultados para formar el mapa de características (feature map).

4. ReLU (Rectified Linear Unit): Se aplica la activación ReLU para eliminar valores negativos.



📚 Ejemplo en Keras

```python
# Definir una capa convolucional en Keras
conv_layer = Conv2D(filters=32, kernel_size=(3,3), activation='relu', input_shape=(28,28,1))
```

###🔎 2. Capa de Agrupación (Pooling Layer)

📊 Concepto

Las capas de agrupación (pooling layers) reducen la dimensionalidad de los mapas de características, mejorando la eficiencia y evitando el sobreajuste.

🔢 Tipos de Pooling

1. Max Pooling (Agrupación Máxima)

Selecciona el valor más alto dentro de una ventana deslizante (2x2, 3x3, etc.).

Ejemplo en Keras:

```python
max_pooling_layer = MaxPooling2D(pool_size=(2,2))
```

2. Average Pooling (Agrupación Promedio)

En lugar de seleccionar el máximo, toma el promedio de los valores en cada región.

Ejemplo en Keras:

```python
average_pooling_layer = AveragePooling2D(pool_size=(2,2))
```

###🔍 3. Arquitectura de una CNN en Keras

```python
# Definir el modelo en Keras
model = Sequential()

# Capa Convolucional + MaxPooling
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
model.add(MaxPooling2D(pool_size=(2,2)))  # 📈 Reduce el tamaño

# Segunda Capa Convolucional + MaxPooling
model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))  # 📈 Otra reducción

# Aplanamiento y Capas Densas
model.add(Flatten())  # 📈 Convierte el tensor en un vector
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))  # 🔄 10 clases (ejemplo MNIST)

# Compilar el modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Mostrar la arquitectura
model.summary()
```

###💡 4. Conclusiones

✅ La capa convolucional extrae patrones clave de las imágenes.

✅ Las capas de pooling reducen la dimensionalidad y mejoran la eficiencia.

✅ MaxPooling es la opción más usada para preservar las características más importantes.

✅ Las CNNs combinan estas capas para aprender a reconocer imágenes de manera eficiente.

Las CNNs se utilizan ampliamente en reconocimiento de imágenes, detección de objetos, visión por computadora y muchas aplicaciones de IA. 🚀
